```python
data = {
    'Waktu': ['09:00', '11.30–14.00', '14.50–17.00', '17.30–18.30', '18.36-21.00'],
    'Pengunjung': [3, 28, 25, 13, 4]
}


def sequential_search(pengunjung_list, waktu_list, threshold=5):
    hasil = []
    for i, val in enumerate(pengunjung_list):
        if val < threshold:
            hasil.append((waktu_list[i], val))
    return hasil


def binary_search_max(pengunjung_list, waktu_list):

    combined = list(zip(pengunjung_list, waktu_list))
    combined.sort(key=lambda x: x[0])


    left = 0
    right = len(combined) - 1

    while left <= right:
        mid = (left + right) // 2

        left_val = combined[mid - 1][0] if mid - 1 >= 0 else float('-inf')
        mid_val = combined[mid][0]
        right_val = combined[mid + 1][0] if mid + 1 < len(combined) else float('-inf')


        if mid_val >= left_val and mid_val >= right_val:
            return combined[mid][1], mid_val
        elif right_val > mid_val:
            left = mid + 1
        else:
            right = mid - 1

    return combined[-1][1], combined[-1][0]



def titik_ekstrim(pengunjung_list):
    maks = max(pengunjung_list)
    min_val = min(pengunjung_list)
    return maks, min_val


def perbandingan(data):
    print("=== pengunjung < 5 ===")
    hasil_seq = sequential_search(data['Pengunjung'], data['Waktu'])
    if hasil_seq:
        for waktu, val in hasil_seq:
            print(f"Waktu = {waktu}, Pengunjung = {val}")
    else:
        print("✗ Tidak ditemukan")

    print("\n=== waktu dengan pengunjung terbanyak ===")
    waktu_bin, val_bin = binary_search_max(data['Pengunjung'], data['Waktu'])
    print(f"Waktu = {waktu_bin}, Pengunjung = {val_bin}")

    print("\n=== Titik Ekstrim ===")
    maksimum, minimum = titik_ekstrim(data['Pengunjung'])
    print(f"Maksimum Pengunjung: {maksimum}")
    print(f"Minimum Pengunjung: {minimum}")


perbandingan(data)
```

```python
import matplotlib.pyplot as plt

data = {
    'Waktu': ['09:00', '11.30–14.00', '14.50–17.00', '17.30–18.30', '18.36-21.00'],
    'Pengunjung': [3, 28, 25, 13, 4]
}

plt.figure(figsize=(10, 5))
plt.plot(data['Waktu'], data['Pengunjung'], marker='o', linestyle='-', color='blue')
plt.title("Distribusi Jumlah Pengunjung per Waktu")
plt.xlabel("Waktu")
plt.ylabel("Jumlah Pengunjung")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Waktu': ['09:00', '11.30–14.00', '14.50–17.00', '17.30–18.30', '18.36-21.00'],
    'Pengunjung': [3, 28, 25, 13, 4]
}

maks = max(data['Pengunjung'])
min_val = min(data['Pengunjung'])

idx_max = data['Pengunjung'].index(maks)
idx_min = data['Pengunjung'].index(min_val)

bar_colors = ['gray'] * len(data['Pengunjung'])
bar_colors[idx_max] = 'blue'
bar_colors[idx_min] = 'red'

plt.figure(figsize=(10, 5))
sns.barplot(x=data['Waktu'], y=data['Pengunjung'], palette=bar_colors)
plt.title("Nilai Maksimum (Biru) dan Minimum (Merah)")
plt.xlabel("Waktu")
plt.ylabel("Jumlah Pengunjung")
plt.tight_layout()
plt.show()
```
